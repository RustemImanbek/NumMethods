```python
import numpy as np

def gauss_seidel(A, b, x0=None, epsilon=1e-10, max_iterations=1000):

    n = len(A)
    x = x0 if x0 is not None else np.zeros(n) 
    k = 1 
    
    while k <= max_iterations:
        delta = 0  
        for i in range(n):
            s = 0
            for j in range(i):  
                s += A[i][j] * x[j]
            for j in range(i + 1, n):  
                s += A[i][j] * x[j]
            
            new_x = (b[i] - s) / A[i][i] 
            d = abs(new_x - x[i])  
            if d > delta:
                delta = d  
            x[i] = new_x  
        
        if delta < epsilon:
            return x  
        
        k += 1  
    
    raise ValueError("Метод не сошелся за заданное число итераций")

# Пример использования
A = np.array([[4, -1, 0, 0],
              [-1, 4, -1, 0],
              [0, -1, 4, -1],
              [0, 0, -1, 3]], dtype=float)

b = np.array([15, 10, 10, 10], dtype=float)

x0 = np.array([0, 0, 0, 0], dtype=float)

solution = gauss_seidel(A, b, x0)
print("Решение:", solution)


```

```python
Ax_computed = np.dot(A, solution)
print(Ax_computed)

if np.allclose(Ax_computed, b, atol=1e-6):
    print("Решение корректно!")
else:
    print("Решение некорректно!")
```

```python
import numpy as np
from scipy.optimize import bisect

def f(x):
    # Функция, корень которой ищем
    return np.tan(3*x) + np.exp(2.78*x) - 1
#     return x ** 3 - x  -2

def bisection_method(a, b, epsilon):
    if f(a) * f(b) >= 0:
        print("Корень не найден на данном интервале")
        return None
    
    n = 1
    while abs(b - a) >= epsilon:
        xn = (a + b) / 2
        print(f"Итерация {n}: a = {a}, b = {b}, xn = {xn}, f(xn) = {f(xn)}")
        
        if f(xn) == 0:
            return xn  # Точный корень найден
        
        if f(a) * f(xn) < 0:
            b = xn  # Корень находится в левом подотрезке
        else:
            a = xn  # Корень находится в правом подотрезке
        
        n += 1
    
    return (a + b) / 2  # Приближенное значение корня


a0 = -1
b0 = 2
epsilon = 0.001
root = bisection_method(a0, b0, epsilon)
if root is not None:
    print(f"Приближенное значение корня: {root}")
```

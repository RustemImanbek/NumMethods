```python
import numpy as np

def f(x):
    return x**3 - 4*x + 1

def phi(x):
    return (x**3 + 1) / 4  # Выражение для простой итерации

def simple_iteration_method(x0, epsilon=1e-6, max_iterations=100):
    """Метод простой итерации с улучшенным выводом итераций."""
    print("\nМетод простой итерации: поиск корня уравнения f(x) = 0")
    print("{:<10} {:<15} {:<15}".format("Итерация", "x", "f(x)"))
    print("-" * 40)

    for i in range(max_iterations):
        x1 = phi(x0)
        print("{:<10} {:<15.6f} {:<15.6f}".format(i+1, x1, f(x1)))

        if abs(x1 - x0) < epsilon:
            print("-" * 40)
            return x1
        x0 = x1
    
    print("-" * 40)
    return x1

# Начальное приближение
x0 = 1
root_simple_iteration = simple_iteration_method(x0)
print(f"\nПриближенное значение корня (метод простой итерации): {root_simple_iteration}")
```

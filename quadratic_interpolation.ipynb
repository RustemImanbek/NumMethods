```python
import numpy as np

def gauss_elimination_block_scheme(A, b):

    n = len(b)

    # Прямой ход (обнуление нижней части матрицы)
    for k in range(n):
        if A[k, k] == 0:
            for i in range(k+1, n):
                if A[i, k] != 0:
                    A[[k, i]] = A[[i, k]]
                    b[k], b[i] = b[i], b[k]
                    break
            else:
                raise ValueError("Система несовместна или имеет бесконечно много решений")

        for i in range(k+1, n):
            c = A[i, k] / A[k, k]
            for j in range(k, n):  
                A[i, j] -= c * A[k, j]
            b[i] -= c * b[k]  

    # Обратный ход (обратная подстановка)
    x = np.zeros(n)
    for i in range(n-1, -1, -1):
        s = sum(A[i, j] * x[j] for j in range(i+1, n)) 
        x[i] = (b[i] - s) / A[i, i]  

    return x

def quadratic_interpolation(x_values, y_values, x_star):
    
    if len(x_values) < 3:
        raise ValueError("Для квадратичной интерполяции нужно минимум три точки.")


    i = 0
    while i < len(x_values) and x_star >= x_values[i]:
        i += 1

    if i == 0 or i >= len(x_values):  
        raise ValueError(f"x* = {x_star} вне диапазона доступных значений ({x_values[0]} - {x_values[-1]})")


    if i >= 2:
        x_vals = x_values[i - 2:i + 1]
        print(x_vals)
        y_vals = y_values[i - 2:i + 1]
        print(y_vals)
    else:
        x_vals = x_values[:3]
        y_vals = y_values[:3]


    A = np.array([
        [x_vals[0]**2, x_vals[0], 1],
        [x_vals[1]**2, x_vals[1], 1],
        [x_vals[2]**2, x_vals[2], 1]
    ], dtype=float)

    B = np.array(y_vals, dtype=float)


    a, b, c = gauss_elimination_block_scheme(A.copy(), B.copy())


    y_interp = a * x_star**2 + b * x_star + c

    return y_interp
```

```python
# Пример использования
x_values = [0.15, 0.30, 0.40] 
y_values = [2.17, 3.63, 5.07]
x_star = 0.32

y_star = quadratic_interpolation(x_values, y_values, x_star)
print(f"Интерполированное значение в x = {x_star}: y = {y_star}")
```

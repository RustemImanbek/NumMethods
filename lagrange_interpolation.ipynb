```python
import numpy as np

def lagrange_interpolation(x_values, y_values, x):

    n = len(x_values)
    result = 0

    for i in range(n):
        term = y_values[i]
        for j in range(n):
            if i != j:
                term *= (x - x_values[j]) / (x_values[i] - x_values[j])
        result += term

    return result


# x_values = np.array([0, 1, 2, 5])
# y_values = np.array([2, 3, 12, 147])

x_values = [0.15, 0.30, 0.40] 
y_values = [2.17, 3.63, 5.07]


x_target = 0.32
y_target = lagrange_interpolation(x_values, y_values, x_target)


print(f"Значение интерполяционного многочлена в точке x = {x_target}: {y_target}")
```

```python
def linear_interpolation(x_star, x_values, y_values):
    i = 0
    while i < len(x_values) and x_star >= x_values[i]:
        i += 1

    if i == 0 or i >= len(x_values):  
        raise ValueError("x* вне диапазона доступных значений")

    a = (y_values[i] - y_values[i-1]) / (x_values[i] - x_values[i-1])
    b = y_values[i-1] - a * x_values[i-1]


    y_star = a * x_star + b
    return y_star


# x_values = [1, 2, 3, 4, 5]
# y_values = [2, 4, 6, 8, 10]
# x_star = 2.5 

# x_values = [0.5, 1.5, 3.0, 4.2, 5.8, 7.1, 8.9]
# y_values = [1.2, 3.5, 5.1, 7.8, 9.3, 10.5, 12.7]
# x_star = 2.5

x_values = [0.15, 0.30, 0.40, 0.55]
y_values = [2.17, 3.63, 5.07, 7.78]
x_star = 0.32




y_star = linear_interpolation(x_star, x_values, y_values)
print(f"Интерполированное значение y* для x*={x_star}: {y_star}")
```

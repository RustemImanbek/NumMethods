```python
import numpy as np

def gauss_elimination_block_scheme(A, b):

    n = len(b)

    # Прямой ход (обнуление нижней части матрицы)
    for k in range(n):
        if A[k, k] == 0:
            for i in range(k+1, n):
                if A[i, k] != 0:
                    A[[k, i]] = A[[i, k]]
                    b[k], b[i] = b[i], b[k]
                    break
            else:
                raise ValueError("Система несовместна или имеет бесконечно много решений")

        for i in range(k+1, n):
            c = A[i, k] / A[k, k]
            for j in range(k, n):  
                A[i, j] -= c * A[k, j]
            b[i] -= c * b[k]  

    # Обратный ход (обратная подстановка)
    x = np.zeros(n)
    for i in range(n-1, -1, -1):
        s = sum(A[i, j] * x[j] for j in range(i+1, n)) 
        x[i] = (b[i] - s) / A[i, i]  

    return x


A = np.array([[2, 1, -1], [-3, -1, 2], [-2, 1, 2]], dtype=float)
b = np.array([8, -11, -3], dtype=float)


solution = gauss_elimination_block_scheme(A.copy(), b.copy())


print("\n✅ Решение системы:", solution)
```

```python
def euler_method(f, x0, y0, h, n):

    x = x0
    y = y0

    print("Метод Эйлера:")
    print(f"Начальные условия: x0 = {x0}, y0 = {y0}")
    print(f"Шаг h = {h}, Число шагов n = {n}")
    print("-" * 40)
    print(f"{'Шаг':<6}{'x':<10}{'y':<10}")
    print("-" * 40)

    for i in range(n + 1):
        print(f"{i:<6}{x:<10.4f}{y:<10.4f}")
        y = y + h * f(x, y)
        x = x + h

    print("-" * 40)
    print(f"Итоговое значение: x = {x:.4f}, y = {y:.4f}")


def f(x, y):
    return x + y



x0 = 0      
y0 = 1      
h = 0.1     
n = 10      


euler_method(f, x0, y0, h, n)

#216
```

```python
import numpy as np

def simple_iteration_corrected(A, b, x0=None, eps=1e-6, max_iter=1000):
    """
    Решает систему Ax = b методом простой итерации с корректной формулой.
    """

    # 🔹 1. Преобразуем A в разложение D + R
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    D = np.diag(np.diag(A))  # Выделяем диагональную матрицу D
    R = A - D  # Вычисляем матрицу R (все остальные элементы)

    # 🔹 2. Выбираем начальное приближение (если не задано, используем нулевой вектор)
    if x0 is None:
        x0 = np.zeros(len(A))

    x = x0.copy()
    k = 0  # Счётчик итераций

    # 🔹 3. Итерационный процесс
    while True:
        # Вычисляем новое значение по формуле x_(k+1) = D^-1 * (b - R * x_k)
        x_new = np.linalg.inv(D) @ (b - R @ x)

        # Вычисляем ошибку ||x_new - x||
        e = np.max(np.abs(x_new - x))

        # Выводим итерацию
        print(f"🔄 Итерация {k+1}: x = {x_new}, max(|e|) = {e}")

        # Проверяем условие остановки
        if e < eps:
            break

        # Обновляем x и счётчик итераций
        x = x_new
        k += 1

        if k >= max_iter:
            print("⚠️ Превышено максимальное число итераций")
            break

    return x, k

# 🔹 Матрица с диагональным преобладанием (которая должна сходиться)
A = np.array([[10, 1, 1],
              [2, 10, 1],
              [2, 2, 10]])  # Матрица коэффициентов
b = np.array([12, 13, 14])  # Вектор правой части
x0 = np.array([0.0, 0.0, 0.0])  # Начальное приближение

# 🔹 Решаем систему методом простой итерации
solution, iterations = simple_iteration_corrected(A, b, x0)

# 🔹 Проверка: умножаем A на найденный x и сравниваем с b
Ax = np.dot(A, solution)
residual = b - Ax  # Невязка после нахождения решения

# Вывод результатов
print("\n✅ Найденное решение x:")
print(solution)

print(f"\n🔄 Количество итераций: {iterations}")

print("\n✅ Проверка: Ax (должно быть примерно равно b):")
print(Ax)

print("\n✅ Невязка (должна быть близка к нулю):")
print(residual)
```

```python
import numpy as np

def f(x):
    return x**3 - 4*x + 1

def df(x):
    return 3*x**2 - 4

def newton_method(x0, epsilon=1e-6, max_iterations=100):
    """Метод Ньютона (касательных) для нахождения корня с улучшенным выводом итераций."""
    print("\nМетод Ньютона: поиск корня уравнения f(x) = 0")
    print("{:<10} {:<15} {:<15}".format("Итерация", "x", "f(x)"))
    print("-" * 40)
    
    for i in range(max_iterations):
        fx = f(x0)
        dfx = df(x0)
        print("{:<10} {:<15.6f} {:<15.6f}".format(i+1, x0, fx))

        if abs(fx) < epsilon:
            print("-" * 40)
            return x0
        if dfx == 0:
            print("Производная обратилась в ноль, метод Ньютона не применим")
            return None
        x0 = x0 - fx / dfx
    
    print("-" * 40)
    return x0

# Начальное приближение
x0 = 1
root_newton = newton_method(x0)
print(f"\nПриближенное значение корня (метод Ньютона): {root_newton}")
```

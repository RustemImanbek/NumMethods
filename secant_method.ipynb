```python
import numpy as np

def f(x):
    return x**3 - 4*x + 1

def secant_method(a, b, epsilon=1e-6, max_iterations=100):
    """Метод секущих для нахождения корня с улучшенным выводом итераций."""
    if f(a) * f(b) > 0:
        print("Метод секущих не применим: f(a) и f(b) должны иметь разные знаки.")
        return None
    
    print("\nМетод секущих: поиск корня уравнения f(x) = 0")
    print("{:<10} {:<15} {:<15}".format("Итерация", "x", "f(x)"))
    print("-" * 40)

    for i in range(max_iterations):
        c = b - f(b) * (b - a) / (f(b) - f(a))
        print("{:<10} {:<15.6f} {:<15.6f}".format(i+1, c, f(c)))

        if abs(f(c)) < epsilon:
            print("-" * 40)
            return c
        a, b = b, c
    
    print("-" * 40)
    return c

# Определение начального интервала
a, b = 0, 1  # Можно изменить при необходимости
root_secant = secant_method(a, b)
print(f"\nПриближенное значение корня (метод секущих): {root_secant}")
```

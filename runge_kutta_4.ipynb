```python
def runge_kutta_4(f, x0, y0, h, n):

    x_values = [x0]
    y_values = [y0]
    
    x = x0
    y = y0

    print("Метод Рунге–Кутта 4-го порядка:")
    print(f"Начальные условия: x0 = {x0}, y0 = {y0}")
    print(f"Шаг h = {h}, Число шагов n = {n}")
    print("-" * 40)
    print(f"{'Шаг':<6}{'x':<10}{'y':<10}")
    print("-" * 40)

    for i in range(n):
        print(f"{i:<6}{x:<10.4f}{y:<10.4f}")

        k1 = h * f(x, y)
        k2 = h * f(x + h / 2, y + k1 / 2)
        k3 = h * f(x + h / 2, y + k2 / 2)
        k4 = h * f(x + h, y + k3)

        y = y + (k1 + 2*k2 + 2*k3 + k4) / 6
        x = x + h

        x_values.append(x)
        y_values.append(y)

    print(f"{n:<6}{x:<10.4f}{y:<10.4f}")
    print("-" * 40)
    print("Расчёт завершён.")

    return x_values, y_values


def f(x, y):
    return x + y



x0 = 0      
y0 = 1      
h = 0.1     
n = 10      


x_vals, y_vals = runge_kutta_4(f, x0, y0, h, n)


```

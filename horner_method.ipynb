```python
import time

def horner_method_with_i_decrement(n, coefficients, x):
    """
    Оценка многочлена с использованием метода Горнера с уменьшением i на каждом шаге.
    
    Аргументы:
    n (int): Степень многочлена (количество коэффициентов - 1).
    coefficients (list): Список коэффициентов [b_n, b_(n-1), ..., b_0].
    x (float): Значение, в котором оценивается многочлен.
    
    Возвращает:
    float: Результат оценки многочлена в точке x.
    """
    result = coefficients[0]  # Начинаем с первого коэффициента
    
    # Применяем метод Горнера с уменьшением i
    i = 1  # начинаем с b_(n-1)
    while i <= n:
        result = result * x + coefficients[i]
        i += 1  # Увеличиваем i на 1 для движения по списку коэффициентов

    return result


```

```python

# Пример для n = 7:
n = 7
coefficients = [1, -2, 3, -4, 5, -6, 7, -8]
x_value = 2

start_time = time.time()

for _ in range(1000000):  # Повторяем 1 миллион раз
    result = horner_method_with_i_decrement(n, coefficients, x_value)

end_time = time.time()

execution_time = end_time - start_time
print(f"Время выполнения (среднее за 1 итерацию): {execution_time / 1000000:.10f} секунд")
```

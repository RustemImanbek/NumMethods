```python
import numpy as np

# Параметры задачи
a = 3.0             # скорость волны
h = 0.15            # шаг по пространству (в книге — это b)
tau = 0.02          # шаг по времени (в книге — это c)
x_max = 1.0         # граница по x
t_max = 1.0         # граница по t
d = 0.15            # вершина f(x)
l = 0.55            # правая граница f(x)

# Правильная сетка, кратная шагу
x = np.arange(0, x_max + h, h)
t = np.arange(0, t_max + tau, tau)

# Количество узлов
N = len(x) - 1
M = len(t) - 1

# Коэффициент устойчивости
r = (a * tau / h) ** 2

# Начальное отклонение f(x)
def f(x):
    return 3.0 * np.where(x < d, x / d, np.where(x <= l, (l - x) / (l - d), 0.0))

# Инициализация сетки решения
u = np.zeros((M + 1, N + 1))

# Начальное условие: u(x, 0) = f(x)
u[0, :] = f(x)

# Второй временной слой: g(x) = 0, поэтому используем формулу с r/2
u[1, 1:N] = u[0, 1:N] + (r / 2) * (u[0, 2:N+1] - 2 * u[0, 1:N] + u[0, 0:N-1])

# Основной расчёт всех временных слоёв
for j in range(1, M):
    for i in range(1, N):
        u[j+1, i] = 2 * (1 - r) * u[j, i] + r * (u[j, i+1] + u[j, i-1]) - u[j-1, i]

# Вывод результатов
for j in range(M + 1):
    print(f"t = {t[j]:.2f}:", ["{:.3f}".format(val) for val in u[j]])
```
